```python
str1+", "+str3  # 相加
```

## 2.3 字符串切片

<font color="#dd0000">所有的 Python 有序可迭代对象（字符串、列表、元组），都支持切片，遵循"从0开始"、"左闭右开"的原则：

“从0开始”: 第一个对象元素的序号是0，不是1！

“左闭右开”: 切片结果包含左边，不包含右边！

```python
str5 = 'I love Python'
```

<img src="img/切片_字符串.jpg" width=500 align='left'>

```python
str5[2:8]  # 常规切片（从0开始，左闭右开）
```

```python
str5[:8]  # 省略开头位置，默认从0开始
```

```python
str5[2:]  # 省略结尾位置，默认到最后一位
```

```python
str5[:]  # 省略开头和结尾位置，默认选择全部，相当于本身
```

```python
str5[2:-1] #负数代表倒序，最后一位是-1，以此类推
```

## 2.4 字符串的格式化

<font color="#dd0000">格式化的方法，熟练掌握一种即可

```python
"I am {}, and I love {}".format('马哥', 'python')  # 用format方法
```

```python
"I am %s, and I have %d python IDE" % ('马哥', 3)  # 用%操作符
```

```python
who = '马哥'
age = 18
f'I am {who}, my age is {age}'  # 用f-string方法
```

# 五、字典

## 5.1 定义一个字典

<font color="#dd0000">字典用{}定义，是一种键值对的数据类型，键和值用:分隔，对和对用,分隔，键不可重复。

```python
mage = {'name': '马哥', 'age': 18, 'major': 'Python'}  # 定义一个字典
```

```python
type(mage)  # 查看数据类型
```

## 5.2 访问字典中元素

```python
mage['age']
```

## 5.3 修改字典中元素

```python
mage['age'] = 19 #修改元素值
```

```python
mage #查看修改结果
```

## 5.4 字典常用内置函数

```python
len(mage)  # 查看字典长度，即字典中有几个元素
```

```python
'name' in mage  # 判断字典中是否存在指定键
```

```python
mage.items()  # 查看字典中所有元素
```

```python
# 循环读取字典中的键和值
for k, v in mage.items():
    print(k, '->', v)
```

```python
mage.keys()  # 查看所有键
```

```python
mage.values()  # 查看所有值
```

# 六、集合

<font color="#dd0000">集合用{}定义，是一种无序、元素不可重复的数据类型

## 6.1 定义一个集合

```python
school = {'小学', '大学', '高中', '初中', ('幼儿园小班', '幼儿园中班', '幼儿园大班')}
```

```python
type(school)  # 查看数据类型
```

## 6.2 集合常用内置函数

```python
len(school)  # 计算集合长度，即集合元素的个数
```

```python
school.remove('大学')  # 删除集合元素
```

```python
school
```

```python
school.add('大学')  # 增加集合元素
```

```python
school
```

## 6.3 利用集合去重（重要）

<font color="#dd0000">利用集合元素的不可重复特性，将list或其他可迭代数据类型去除重复

```python
programs = ['python', 'shell', 'golang', 'shell', 'java']  # 定义一个列表
```

```python
programs = list(set(programs))  # 把list转换为set，再把set转换为list，实现列表去重
```

```python
programs  # 查看去重结果
```

# END、总结

Python共6种数据类型。其中，

不可变的数据类型有：数值、字符串、元组
