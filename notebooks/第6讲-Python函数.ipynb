# 一、无参数的函数

```python
# 定义函数
def hello1():
    print('我是：马哥python说')
```

```python
# 调用函数
hello1()
```

# 二、有参数的函数

## 2.1 无默认值参数的函数

```python
# 定义函数
def hello2(v_name):
    print('我是：{}'.format(v_name))
```

```python
# 调用函数
hello2('马哥python说')
```

## 2.2 有默认值参数的函数

```python
# 定义函数，函数的v_major参数的默认值是Python
def hello3(v_name, v_major='Python'):
    print('我是{}, 我的专业是{}'.format(v_name, v_major))
```

```python
# 定义函数时，如果存在默认值参数，默认值参数必须放在后面，否则会报错，如下：
def hello4(v_major='Python', v_name):
    print('我是{}, 我的专业是{}'.format(v_name, v_major))
```

```python
# 不传参数
hello3('马哥')
```

```python
# 传入指定参数值
hello3('马哥', 'Java')
```

## 2.3 显性传参

<font color="#dd0000">上面举的例子，都是隐性传参，即，调用函数时，没指明参数名称；

<font color="#dd0000">下面介绍显性传参，即，调用函数时，指明参数名称。

```python
# 定义函数
def hello5(v_name, v_major):
    print('我是{}, 我的专业是{}'.format(v_name, v_major))
```

```python
# 显性传参时，指明参数名称，参数可以打乱顺序
hello5(v_major='Python', v_name='马哥')
```

# 三、有返回值的函数

<font color="#dd0000">python中用"return"这个关键字返回函数值

## 3.1 一个返回值

```python
# 求两个数的最大值
def f_max(num1, num2):
    if num1 > num2:
        return num1
    else:
        return num2
```

```python
max_value = f_max(num1=12, num2=9)
```

```python
max_value
```

## 3.2 多个返回值

```python
# 计算圆的周长和面积
def f_calc_circle(r):
    """计算圆的周长和面积，r代表半径"""
    length = 2*3.14*r  # 周长
    area = 3.14*(r**2)  # 面积
    return length, area
```

```python
length, area = f_calc_circle(r=3)
```

```python
length
```

```python
area
```

# 四、匿名函数

<font color="#dd0000">python中用"lambda"这个关键字定义匿名函数：（lambda 参数：表达式）冒号前面是参数，冒号后面是表达式返回值
    
<font color="#dd0000">匿名函数一般用于实现代码较短、逻辑简单的功能，一行代码完成函数定义

```python
f_sum=lambda x, y: x+y
```

```python
f_sum(1, 2)
```
