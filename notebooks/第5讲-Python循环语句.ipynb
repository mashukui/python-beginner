# 一、for循环

## 1.1 普通for循环

```python
list1 = ['Python', 'C++', 'Java', 'PHP']  # 定义一个列表
```

```python
# 利用for循环遍历列表元素
for i in list1:
    print(i)
```

## 1.2 有else的for循环

```python
# 有else的for循环
for i in list1:
    print(i)
else:  # 如果for循环遇到break，则不会执行else体；如果for循环正常结束，则会执行else体
    print('正常结束！')
```

## 1.3 有break的for循环

```python
list1
```

<font color="#dd0000">break：跳出整个for循环

```python
# break语句
for i in list1:
    print('当前编程语言是:{}'.format(i))
    if i == 'Java':
        print('this is Java, break now!')
        break  # 如果for循环遇到break，则跳出循环体
else:
    print('正常结束！')
```

## 1.4 有continue的for循环

<font color="#dd0000">continue：跳出本轮循环

```python
list1
```

```python
# continue语句
for i in list1:
    if i == 'Java':
        continue  # 如果遇到continue，跳过当前循环的剩余语句，然后继续进行下一轮循环
    print('我最喜欢{}了'.format(i))
```

## 1.5 用for循环遍历range函数

```python
# 用for循环遍历range函数
for i in range(5):
    print('当前序号是:{}'.format(i))
```

## 1.6 嵌套for循环

1.6.1 嵌套for循环打印上学的时期

```python
# 嵌套for循环打印上学的时期
schools = ['小学', '初中', '高中']
terms = ['上学期', '下学期']
for school in schools:  # 外层循环
    for term in terms:  # 内层循环
        print('我在上{}的{}'.format(school, term))
```

1.6.2 嵌套for循环遇到break时，<font color="#dd0000">跳出最内层循环</font>，不是跳出所有循环：

```python
# 嵌套for循环打印上学的时期-有break
schools = ['小学', '初中', '高中']
terms = ['上学期', '下学期']
for school in schools:  # 外层循环
    for term in terms:  # 内层循环
        if school == '初中' and term == '上学期':
            break
        print('我在上{}的{}'.format(school, term))
```

1.6.3 嵌套for循环打印九九乘法表

```python
# 嵌套for循环打印九九乘法表
for i in range(1, 10):
    for j in range(1, 10):
        if j <= i:  # 第几层打印几个
            result = i*j
            print('{}*{}={} '.format(j, i, result), end='')  # 打印完不换行
    print()  # 打印换行
```

# 二、while循环

## 2.1 普通while循环

```python
a = 0  # 初始化
```

```python
while a < 3:
    print('a is:', a)
    a += 1  # 等价于：a=a+1
```

## 2.2 有if和break的while循环

```python
a = 0  # 初始化
```

```python
while True:
    print('a is:', a)
    a += 1
    if a > 3:
        break
```

## 2.3有else的while循环

<font color="#dd0000">当不满足while后面的条件时，执行else语句；如果遇到break，则不执行else语句

```python
a = 0  # 初始化
```

```python
while a < 3:
    print('a is:', a, ', a<3')
    a += 1
else:
    print('a>=3了')
```

## 2.4 无限循环/死循环

<font color="#dd0000">无限循环，即，死循环，程序永远无法跳出循环体，开发人员应该避免这种情况发生！
    
<font color="#dd0000">解决办法：循环体中增加if...break的语句

```python
# 执行此代码，会进入无限循环，可点击工具栏"中断服务"按钮，停止程序！
while 1 == 1:
    input('请输入任意内容：')
```

```python
# 执行此代码，会进入无限循环，可点击工具栏"中断服务"按钮，停止程序！
while True:
    input('请输入任意内容：')
```

# 三、总结

## 3.1 关于两种循环的应用

for循环：一般用于遍历读取可迭代对象的元素，如列表、元组、字典等。

while循环：一般用于反复执行语句，满足一定条件退出循环。

## 3.2 关于break和continue

break：跳出循环体（当存在嵌套循环时，跳出最内层循环体）

continue：跳出本轮循环，即，不再执行本轮循环中continue后面的语句
