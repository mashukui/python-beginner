# 一、安装Python解释器

python官网下载地址：https://www.python.org/downloads/

1、选择需要安装的python版本，在安装地址中，选择Python 3.8.5，点击Download：

<img src="img/1.png" width=1000 align='left'>

跳转到https://www.python.org/downloads/release/python-385/
然后选择对应的版本，进行下载安装。

<img src="img/2.png" width=1000 align='left'>

2、以Windows用户为例，下载好后，双击exe文件安装，记得勾选"Add Python3.8 to Path"，这样安装好之后就不用配置环境变量了。

<img src="img/3.png" width=1000 align='left'>

3、安装完成之后，在cmd命令行里执行python命令，出现以下界面，证明安装成功了。

<img src="img/4.png" width=1000 align='left'>

# 二、安装PyCharm

python的集成开发环境IDE有很多种，初学者建议采用PyCharm作为入门工具。

PyCharm官网下载地址：https://www.jetbrains.com/pycharm/download

PyCharm有2个版本，专业版和社区版，初学者选择社区版就够用了。以Windows为例，点击Download下载：

<img src="img/5.png" width=1000 align='left'>

下载完成之后，双击exe文件开始安装，直接点下一步，直至安装完成就OK了。

安装完成之后，打开PyCharm，新建一个项目，在设置项里：

<img src="img/6.png" width=1000 align='left'>

新建一个python文件(左侧目录树，右键New->Python File)，输入HelloWorld代码，运行，查看结果，打印成功。

<img src="img/7.png" width=1000 align='left'>
