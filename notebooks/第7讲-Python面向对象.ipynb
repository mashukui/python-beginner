<font color="#dd0000">面向对象是一种编程思想，它把一类有共同特征的事物抽象成一个类（class），把其共有属性和共有方法封装到类里面。

# 一、类的定义和调用

<font color="#dd0000">类的定义有三种方式:
>不带()：class Animal:  
>带()：class Animal():  
>带(object)：class Animal(object):

```python
# 定义一个动物的类
class Animal():
    def __init__(self, arg_name, arg_age):  # 构造方法，在类实例化时会自动调用
        self.name = arg_name  # 姓名
        self.age = arg_age  # 年龄

    # 定义一个类方法
    def eat(self):  # 吃饭
        print(self.name + "正在吃饭")

    # 定义一个类方法
    def sleep(self):  # 睡觉
        print(self.name + "正在睡觉")
```

```python
# 实例化一个对象
xiaobai = Animal(arg_name='小白', arg_age=3)
```

```python
# 调用类的属性
print('它的年龄是{}岁'.format(xiaobai.age))
```

```python
# 调用类方法
xiaobai.sleep()
```

## 2.1 单继承1：Horse类继承自Animal类

```python
# 定义一个马的类，继承自Animal类
class Horse(Animal):
    def __init__(self, arg_name, arg_age, arg_max_distance):  # 构造方法，在类实例化时会自动调用
        Animal.__init__(self, arg_name, arg_age)  # 调用父类的构造函数
        self.max_distance = arg_max_distance  # 最大奔跑距离

    def run(self):
        print(self.name + '正在奔跑')

    def sleep(self):  # 方法重写
        print(self.name + "正在马棚里睡觉")
```

```python
horse1 = Horse(arg_name='白龙马', arg_age=5, arg_max_distance=20)
```

```python
horse1.name  # 调用继承自父类的属性
```

```python
horse1.max_distance  # 调用子类的属性
```

```python
horse1.run()  # 调用子类的方法
```

```python
horse1.sleep()  # 调用子类的重写方法
```

## 2.2 单继承2：Donkey类继承Animal类

```python
# 定义一个驴的类，继承自Animal类
class Donkey(Animal):
    def __init__(self, arg_name, arg_age, arg_work_hours):  # 构造方法，在类实例化时会自动调用
        Animal.__init__(self, arg_name, arg_age)  # 调用父类的构造函数
        self.work_hours = arg_work_hours  # 工作时长

    def work(self):
        print(self.name + '正在拉磨')
```

```python
donkey1 = Donkey(arg_name='小毛驴', arg_age='3', arg_work_hours=5)
```

```python
donkey1.name  # 调用继承自父类的属性
```

```python
donkey1.work_hours  # 调用子类的属性
```

```python
donkey1.eat()  # 调用继承自父类的方法
```

## 2.4 多重继承

<img src="img/多重继承1.png" width=500 align='left'>

```python
class Mule(Horse, Donkey):
    def play(self):
        print(self.name+'正在玩耍')
```

```python
mule1 = Mule(arg_name='小骡子', arg_age=4, arg_max_distance=20)
```

```python
mule1.run()  # 继承自Horse类的方法
```

```python
mule1.work()  # 继承自Donkey类的方法
```

```python
mule1.play()  # 本类的方法
```

# 三、类的私有属性

```python
class Movie():
    def __init__(self, arg_name, arg_year):
        self.name = arg_name  # 电影名称
        self.__year = arg_year  # 上映年份

    def __get_info(self):  # 私有方法
        print('电影<{}>是{}年上映的'.format(self.name, self.__year))
```

```python
movie1 = Movie(arg_name='大话西游', arg_year='1995')
```

```python
movie1.__year  # 私有属性不可访问，所以报错了
```

```python
movie1.__get_info()  # 私有方法不可访问，所以报错了
```

```python
movie1.name
```
