# 二、几种if句式

```python
age = 62  # 年龄1
```

## 2.1 句式1：if

```python
if age > 60:
    print('老年')
```

## 2.2 句式2：if-else

```python
if age > 60:
    print('老年')
else:
    print('不是老年')
```

## 2.3 句式3：if-elif-else

```python
if 0 < age < 18:
    print('少年')
elif 18 <= age < 35:
    print('青年')
elif 35 <= age < 60:
    print('中年')
else:
    print('老年')
```

## 2.4 句式4：if-and（多条件并且）

```python
if age > 60 and age < 89:  # 并且关系用and表示
    print('低龄老年')
else:
    print('未知')
```

## 2.5 句式5：if-or（多条件或者）

```python
if 1 == 2 or 3 == 3:  # 或者关系用or表示
    print('相等')
else:
    print('不相等')
```

## 2.6 句式6：嵌套if

```python
if 0 < age < 18:
    if 0 < age <= 6:  # 二级if语句
        print('幼儿')
    else:
        print('少年')
elif 18 <= age < 35:
    print('青年')
elif 35 <= age < 60:
    print('中年')
else:
    if 60 <= age < 89:  # 二级if语句
        print('低龄老年')
    else:
        print('高龄老年')
```

# 三、经验总结

<font color="#dd0000">在写if语句时，务必要考虑全面，逻辑清晰，把所有情况分支合理加入到if语句中，避免漏掉、有重叠的情况。

```python
age3 = 18
```

```python
# 漏掉了考虑边界值，导致出现非预期结果！
if 0 < age3 < 18:
    print('少年')
elif 18 < age3 < 35:
    print('青年')
elif 35 <= age3 < 60:
    print('中年')
else:
    print('老年')
```

```python
# 分支出现重叠部分，进入第一个分支语句！
if 0 < age3 < 19:
    print('少年')
elif 17 < age3 < 35:
    print('青年')
elif 35 <= age3 < 60:
    print('中年')
else:
    print('老年')
```
